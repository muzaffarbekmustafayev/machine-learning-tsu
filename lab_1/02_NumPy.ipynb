### Задание 1. Одномерный массив в 3x3, сумма и умножение на 2

```
import numpy as np

arr = np.arange(9).reshape(3, 3)
total = arr.sum() * 2
print("Матрица 3x3:\n", arr)
print("Сумма элементов * 2:", total)
```

### Задание 2. Матрица 4x4, диагональ, срез и сумма

```
arr = np.zeros((4, 4))
np.fill_diagonal(arr, 1)
# Удаляем первую строку и последний столбец
reduced = arr[1:, :-1]
print("Оставшийся массив:\n", reduced)
print("Сумма оставшихся элементов:", reduced.sum())
```

### Задание 3. Случайная матрица 10x10, сортировка и замена минимумов

```
np.random.seed(42)
arr = np.random.randint(10, 99, size=(10, 10))
arr_sorted = np.sort(arr, axis=1)

min_vals = arr_sorted[:, 0].copy()
max_vals = arr_sorted[:, -1].copy()
arr_sorted[:, 0] = max_vals
changes = np.sum(min_vals != max_vals)

print(f"Количество произведенных замен: {changes}")
print("Первые 2 строки матрицы после замен:\n", arr_sorted[:2])
```

### Задание 4. Замена минимума на максимум, умножение на 2 и сортировка

```
arr = np.array([15, 3, 42, 8, 91, 12, 67])
print("Исходный:", arr)

min_idx = np.argmin(arr)
arr[min_idx] = arr.max()
arr = arr * 2
arr_desc = np.sort(arr)[::-1]
print("После преобразования и сортировки по убыванию:", arr_desc)
```

### Задание 5. Объединение массивов (hstack, vstack) и среднее по столбцам

```
a = np.ones((2, 3), dtype=int) * 2
b = np.ones((2, 3), dtype=int) * 5
h_merged = np.hstack((a, b))

c = np.ones((1, 6), dtype=int) * 9
v_merged = np.vstack((h_merged, c))
sorted_rows = np.sort(v_merged, axis=1)
col_means = sorted_rows.mean(axis=0)

print("Итоговый массив:\n", sorted_rows)
print("Средние значения по столбцам:", col_means)
```

### Задание 6. Добавление среднего значения и сортировка строк по новому среднему

```
arr = np.array([[1.0, 4.0, 7.0], [2.0, 3.0, 5.0], [9.0, 8.0, 6.0]])
overall_mean = arr.mean()
arr_mod = arr + overall_mean

row_means = arr_mod.mean(axis=1)
sorted_by_mean = arr_mod[np.argsort(row_means)]

print("Среднее массива:", round(overall_mean, 2))
print("Новые средние по строкам:", np.round(row_means, 2))
print("Массив, отсортированный по средним строк:\n", np.round(sorted_by_mean, 2))
```

### Задание 7. Поиск отрицательных элементов и замена на модуль

```
arr = np.array([-5, 12, -3, 0, 7, -8, 15, -1])
neg_idx = np.where(arr < 0)[0]
orig_sum = arr.sum()

arr_abs = arr.copy()
arr_abs[neg_idx] = np.abs(arr_abs[neg_idx])
new_sum = arr_abs.sum()

print("Индексы отрицательных:", neg_idx)
print(f"Сумма исходных: {orig_sum}, Сумма после замены на модуль: {new_sum}")
```

### Задание 8. Преобразование 1D в 2D, flipud/fliplr и проверка

```
arr1d = np.arange(1, 13)
arr2d = arr1d.reshape(3, 4)

flipped = np.flipud(np.fliplr(arr2d))
restored = np.flipud(np.fliplr(flipped))

print("Исходный 2D:\n", arr2d)
print("Перевернутый:\n", flipped)
print("Совпадает с исходным после возврата:", np.array_equal(arr2d, restored))
```

### Задание 9. Извлечение каждого третьего элемента

```
arr = np.arange(1, 21)
step_arr = arr[::3]
diff = np.setdiff1d(arr, step_arr)

print("Исходный массив:", arr)
print("Каждый 3-й элемент:", step_arr)
print("Разница (оставшиеся элементы):", diff)
```

### Задание 10. Проверка наличия положительных чисел и обработка

```
arr = np.array([-4, -2, 3, 0, 8, -1])

if (arr > 0).any():
    arr = np.where(arr > 0, arr * 2, arr)
else:
    arr = arr + 1

zeros_count = np.count_nonzero(arr == 0)
print("Преобразованный массив:", arr)
print("Количество нулевых элементов:", zeros_count)
```

### Задание 11. Сортировка, разбиение на 4 части и замена минимумов

```
np.random.seed(10)
arr = np.sort(np.random.randint(10, 80, size=16))
parts = np.split(arr, 4)

modified = []
for p in parts:
    p_copy = p.copy()
    p_copy[p_copy == p_copy.min()] = p_copy.max()
    modified.append(p_copy)

result = np.concatenate(modified)
print("Исходный отсортированный:", arr)
print("Результат после замены min на max в каждой из 4 частей:\n", result)
```

### Задание 12. Умножение, округление, уникальные значения

```
arr = np.array([1.2, 2.7, 3.1, 2.7, 4.4, 1.2])
factor = 2.0
scaled = np.round(arr * factor).astype(int)
unique_sorted = np.sort(np.unique(scaled))

print("После умножения и округления:", scaled)
print("Уникальные по возрастанию:", unique_sorted)
```

### Задание 13. Транспонирование, удаление строки/столбца и сравнение

```
arr = np.arange(20).reshape(4, 5)
transposed = arr.T
# Удаляем вторую строку (индекс 1) и третий столбец (индекс 2)
cut = np.delete(transposed, 1, axis=0)
cut = np.delete(cut, 2, axis=1)
restored = cut.T

print("Форма исходного:", arr.shape)
print("Форма после удалений и повторного транспонирования:", restored.shape)
print("Полученный массив:\n", restored)
```

### Задание 14. Неповторяющиеся элементы и замена их на среднее

```
arr = np.array([4, 1, 7, 1, 9, 4, 3, 8, 2, 8])
vals, counts = np.unique(arr, return_counts=True)
uniques = vals[counts == 1]
mean_unique = uniques.mean()

sorted_arr = np.sort(arr).astype(float)
for u in uniques:
    sorted_arr[sorted_arr == u] = mean_unique

print(f"Неповторяющиеся элементы: {uniques} (кол-во: {len(uniques)})")
print(f"Среднее неповторяющихся: {mean_unique:.2f}")
print("Итоговый массив:", sorted_arr)
```

### Задание 15. Разделение по среднему, объединение и медиана

```
arr = np.array([12, 45, 23, 67, 34, 89, 10, 50])
mean_val = arr.mean()

part_low = arr[arr < mean_val]
part_high = arr[arr >= mean_val]
rejoined = np.sort(np.concatenate([part_low, part_high]))
median_val = np.median(rejoined)

print(f"Среднее: {mean_val:.2f}")
print("Объединенный и отсортированный:", rejoined)
print(f"Медиана: {median_val}")
```

### Задание 17. Замена максимумов строк на минимумы

```
arr = np.array([[10, 20, 15], [35, 5, 25], [12, 48, 30]])
old_maxs = arr.max(axis=1)

arr_mod = arr.copy()
for i in range(len(arr_mod)):
    arr_mod[i, np.argmax(arr_mod[i])] = arr_mod[i].min()

new_maxs = arr_mod.max(axis=1)
diff = new_maxs - old_maxs

print("Исходная матрица:\n", arr)
print("Старые максимумы строк:", old_maxs)
print("Новые максимумы строк:", new_maxs)
print("Разность (новые - старые):", diff)
```

### Задание 18. Произведение главной диагонали и сортировка строк

```
arr = np.array([[2, 1, 3], [1, 2, 1], [3, 2, 2]])
diag_prod = np.prod(np.diag(arr))
arr_added = arr + diag_prod
arr_sorted = np.sort(arr_added, axis=1)

print(f"Произведение элементов главной диагонали: {diag_prod}")
print("Массив после прибавления и сортировки строк:\n", arr_sorted)
```

### Задание 19. Треугольная матрица, поворот на 180° и максимальная сумма строк

```
tri = np.triu(np.arange(1, 17).reshape(4, 4))
rot180 = np.rot90(tri, 2)
sum_mat = tri + rot180
max_row_sum = sum_mat.sum(axis=1).max()

print("Треугольная матрица:\n", tri)
print("Сумма с повернутой на 180°:\n", sum_mat)
print("Максимальная сумма по строкам:", max_row_sum)
```

### Задание 20. Сравнение двух массивов и замена различающихся элементов

```
a = np.array([10, 25, 30, 45, 50])
b = np.array([10, 20, 30, 40, 50])

diff_indices = np.where(a != b)[0]
means = (a[diff_indices] + b[diff_indices]) / 2.0

a_new = a.astype(float).copy()
a_new[diff_indices] = means

print("Индексы различий:", diff_indices)
print("Исходный A:", a)
print("Обновленный A:", a_new)
print("Разность между новым и старым A:", a_new - a)
```

### Задание 21. Применение функции к элементам, сумма четных и нечетных

```
arr = np.arange(1, 11)
transformed = arr ** 2

evens = transformed[transformed % 2 == 0]
odds = transformed[transformed % 2 != 0]

print("Преобразованный массив (квадраты):", transformed)
print(f"Сумма четных: {evens.sum()}, Сумма нечетных: {odds.sum()}")
```

### Задание 22. Поворот на 90° против часовой стрелки и обнуление границ

```
arr = np.arange(1, 17).reshape(4, 4)
rot90 = np.rot90(arr, 1)

# Обнуляем границы
rot90[0, :] = 0
rot90[-1, :] = 0
rot90[:, 0] = 0
rot90[:, -1] = 0

restored = np.rot90(rot90, -1)
print("Результат после поворота обратно:\n", restored)
```

### Задание 26. Проверка целых чисел, возведение в куб и сумма

```
arr = np.array([2, 3, 4, 5])
is_all_int = np.issubdtype(arr.dtype, np.integer)
cubed = arr ** 3

diff_sum = cubed.sum() - arr.sum()
print("Массив состоит только из целых?:", is_all_int)
print(f"Сумма исходных: {arr.sum()}, Сумма кубов: {cubed.sum()}")
print(f"Разность сумм: {diff_sum}")
```

### Задание 28. Замена первого положительного элемента на квадратный корень

```
arr = np.array([-8, -2, 16, 25, -3])
pos_indices = np.where(arr > 0)[0]

first_pos = pos_indices[0]
arr_res = arr.astype(float).copy()
arr_res[first_pos] = np.sqrt(arr_res[first_pos])
arr_sorted = np.sort(arr_res)

print("Исходный:", arr)
print(f"Первый положительный индекс: {first_pos} (значение: {arr[first_pos]})")
print("После замены на корень и сортировки:", arr_sorted)
```
