### Задание 1. Вычисление суммы ряда Фибоначчи

```
def fibonacci_sum(limit):
    a, b = 0, 1
    total = 0
    while a <= limit:
        total += a
        a, b = b, a + b
    return total

limit = 50
print(f"Сумма чисел Фибоначчи до {limit}: {fibonacci_sum(limit)}")
```

### Задание 2. Генерация простых чисел

```
def generate_primes(n):
    primes = []
    for num in range(2, n + 1):
        if all(num % i != 0 for i in range(2, int(num ** 0.5) + 1)):
            primes.append(num)
    return primes

print("Простые числа от 2 до 30:", generate_primes(30))
```

### Задание 5. Поиск в списке строк из файла

```
def search_word_in_file(filename, word):
    with open(filename, "r", encoding="utf-8") as f:
        lines = [line.strip() for line in f if line.strip()]
    matching = [l for l in lines if word.lower() in l.lower()]
    return len(matching), matching

count, found_lines = search_word_in_file("data.txt", "learning")
print(f"Найдено строк со словом 'learning': {count}")
for l in found_lines:
    print(" -", l)
```

### Задание 13. Перевод строки в обратный порядок и палиндром

```
def reverse_string(s):
    return s[::-1]

def is_palindrome(s):
    cleaned = "".join(s.lower().split())
    return cleaned == cleaned[::-1]

print("Реверс 'привет':", reverse_string("привет"))
print("'топот' палиндром?:", is_palindrome("топот"))
print("'машина' палиндром?:", is_palindrome("машина"))
```

### Задание 16. Использование функции is_prime

```
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

primes_to_100 = [x for x in range(2, 101) if is_prime(x)]
print("Простые числа от 2 до 100:", primes_to_100)
```

### Задание 18. Обработка исключений

```
def divide_numbers(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Ошибка: деление на ноль! Введите число, отличное от нуля.")
        return None

print("12 / 3 =", divide_numbers(12, 3))
print("12 / 0 =", divide_numbers(12, 0))
```

### Задание 19. Модули и random (Угадай число)

```
import random

def simulate_guessing(secret, guesses):
    for g in guesses:
        if g < secret:
            print(f"Ввод {g}: Загаданное число больше")
        elif g > secret:
            print(f"Ввод {g}: Загаданное число меньше")
        else:
            print(f"Ввод {g}: Точно! Число угадано.")
            break

secret_num = random.randint(1, 50)
print(f"Случайное число сгенерировано. Демонстрация логики:")
simulate_guessing(secret_num, [10, 40, 25, secret_num])
```

### Задание 24. Поиск простых чисел в диапазоне от M до N

```
def primes_in_range(m, n):
    res = []
    for num in range(max(2, m), n + 1):
        if all(num % i != 0 for i in range(2, int(num ** 0.5) + 1)):
            res.append(num)
    return res

print("Простые числа в диапазоне [15, 45]:", primes_in_range(15, 45))
```

### Задание 25. Первые N чисел Фибоначчи

```
def get_fibonacci_n(n):
    fib = []
    a, b = 0, 1
    for _ in range(n):
        fib.append(a)
        a, b = b, a + b
    return fib

print("Первые 12 чисел Фибоначчи:", get_fibonacci_n(12))
```

### Задание 28. Объединение словарей с суммированием

```
dict1 = {'a': 10, 'b': 20, 'c': 30}
dict2 = {'b': 5, 'c': 15, 'd': 25}

merged = dict1.copy()
for k, v in dict2.items():
    merged[k] = merged.get(k, 0) + v

print("Объединенный словарь:", merged)
```
